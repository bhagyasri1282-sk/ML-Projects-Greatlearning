### 2.1 What proportion of the gunny bags have a breaking strength of less than 3.17 kg per sq cm?

- To calculate the proportion of gunny bags with breaking strength less than 3.17kg(x), we need to calculate the `Z-score` of the data.
- `Z-score` defines how far the data point(i.e., 3.17kg) is from the mean w.r.t standard deviation.
- x = 3.17
> $Z-score$ = $\frac{x - \mu}{\sigma}$ = $\frac{3.17 - 5}{1.5}$ = $-1.22$

### Insights:
- Z score of -1.22 is 11.12%. 
- This means `11.12%` of the gunny bags have a breaking strength of less than 3.17kg/sqcm in the data.
- This data can be visually represented as below.

```python
# 2.1 What proportion of the gunny bags have a breaking strength of less than 3.17 kg per sq cm?(Visual representation):
# 1.Plotting the Normal distribution:

# import the required function
from scipy.stats import norm

# 2.Defining Mean and standard deviation
mu, sigma = (5, 1.5)  

# 3. Generating x-values and calculating pdf - probability density function 
x = np.linspace(0, 10, 100) #It consists of values ranging from 0 to 10
y = norm.pdf(x, mu, sigma) #Computing the pdf as it gives a normal curve

# 4. Ploting the Normal distribution curve
plt.figure(figsize=(6, 4)) #Setting up the size of the graph
plt.plot(x, y, color="blue") #plotting the curve by a blue line

# 5. Shade the region where breaking strength < 3.17
x_fill = np.linspace(0, 3.17, 50) #It consists of values ranging from 0 to 3.17
y_fill = stats.norm.pdf(x_fill, mu, sigma) #Computing the pdf for this range of values
plt.fill_between(x_fill, y_fill, color = 'red') #Filling the region between the X_fill and y_fill with red color

# 6. Adding a vertical line at x=3.17
plt.axvline(3.17, color='red',alpha = 0.5, label = 'Area <3.17')
plt.xlabel("Breaking Strength (kg/sq. cm)")
plt.ylabel("Density")
plt.title("Distribution of Gunny Bag Strength")
plt.legend()
plt.show()
```

### 3.1 Zingaro has reason to believe that the unpolished stones may not be suitable for printing. Do you think Zingaro is justified in thinking so?

***The Null and alternative hypotheses can be formulated as:***

> $H_0$ : Unpolished stones are suitable for printing.\
> $H_a$ : Unpolished stones are not suitable for printing.

### Assumptions:
- Data is assumed to be normally distributed
- Samples are independent
- Samples show unequal variances

### Hypothesis test can be used: Levene's test for variance
- Since the question is about checking the suitability of both the type of stones for printing, and when we see the variances or standard deviations are different, we can check how significantly they are different using the Levene's test.

```python
from scipy import stats

# Perform the Levene's test

statistic,p_value = stats.levene(zingaro['Unpolished '], zingaro['Treated and Polished'])
print('p_Value is :', p_value)

if p_value < 0.05:
    print("Inference: Unpolished stones may not be suitable for printing(Significance<0.05)")
else:
    print("Inference: Unpolished stones are suitable for printing(Significance>0.05)")
```

### Insight:
Since p value is much lower than significance, we reject null hypothesis of homegenity of variances

### 3.2 Is the mean hardness of the polished and unpolished stones the same?

***The Null and alternative hypotheses can be formulated as:***

> $H_0$ : $\mu_1 = \mu_2$ - Mean hardness of polished stones is same as unpolished stones\
> $H_a$ : $\mu_1 \neq \mu_2$ - Mean hardness of polished stones is different to unpolished stones

### Assumptions:
- Samples are independent - unpolished and polished
- Normal distributed
- Homogenous variances
- Continuous data

### Hypothesis test can be used: Two sample ttest
- Two sample ttest is used here to compare two independent groups and their means.

```python
from scipy import stats

# Perform the two sample t-test
t_stat, p_value = stats.ttest_ind(zingaro["Unpolished "].dropna(), zingaro["Treated and Polished"].dropna(), equal_var=False)
print("P-value:",p_value)

if p_value < 0.05:
    print("Inference: Mean hardness of polished stones is different to unpolished stones.")
else:
    print("Inference: Mean hardness of polished stones is same as unpolished stones.")
```

### Insight:
As the p-value is 0.0015 which is much lower than the level of significance, so We reject null hypothesis.
Hence we have enough statistical evidence to say that hardness of unpolished stones is not same as polished stones.

### 4.2 How does the hardness of implants vary depending on methods?

***Let's write the null and alternative hypothesis***

>$H_0$: Variance of implants hardness is equal across different methods\
>$H_a$: Variance of implants hardness differs across different methods

****Checking the assumptions:****

- Assumption1 - For testing of normality, `Shapiro-Wilk’s test` is applied to the response variable - this is already performed in 4.1 question
- Assumption2 - For equality of variance, `Levene test` is applied to the response variable 'Methods'.
- Assumption3 - Normally distributed data

```python
# Create groups for Methods column as there are 3 groups in the data
methods = dental["Method"].unique()  # Extracts unique method names
groups = [dental.loc[dental["Method"] == m, "Response"] for m in methods]

# Perform Levene's Test
statistic, p_value = stats.levene(*groups)
print('P-value is:',p_value)

# Interpretation
if p_value < 0.05:
    print("Variances are significantly different across methods(<0.05).")
else:
    print("Variances are similar across methods(>0.05.")
```

*****Insights:******
- Since the p-value is less than the 0.05 significance level, we reject the null hypothesis and variances are significantly different across methods.
- Let us proceed and conduct one way ANOVA test on the data and compute p_value using f_oneway() function.

```python
# Conducting ONE WAY ANOVA TEST:

# Create groups for Methods column as there are 3 groups in the data
groups = [dental.loc[dental["Method"] == m, "Response"] for m in methods]

# Perform f_oneway()'s Test i.e., ANOVA test
statistic, p_value = stats.f_oneway(*groups)
print('P-value is:',p_value)

if p_value < 0.05:
    print("Insight for Q(4.2): We reject the null hypothesis and at least one method has a significantly different mean hardness.")
else:
    print("Insight for Q(4.2): We fail to reject the null hypothesis and implant hardness does not significantly vary across methods.")
```

### 4.3 What is the interaction effect between the dentist and method on the hardness of dental implants for each type of alloy?

- Since we are asked to compare the mean hardness of implants between two factors, i.e., dentist and method with respect to each type of alloy, it is a two way ANOVA test.
- The pairwise_tukeyhsd() function of Statsmodels will be used to compute the test statistic and p-value.

***Let's write the null and alternative hypothesis***

>$H_0$: No significant interaction effect between dentists and methods on the implants hardness for alloy type\
>$H_a$: Significant interaction effect observed between dentists and methods on the implants hardness for alloy type

```python
# Perform Two-Way ANOVA for each alloy separately
for alloy_type in dental['Alloy'].unique():
    print(f"\nTwo-Way ANOVA for Alloy Type {alloy_type}:")
    
    # Filter data for the current alloy type
    subset_data = dental[dental['Alloy'] == alloy_type]
    
    # Fit Two-Way ANOVA model (Dentist × Method)
    model = ols('Response ~ C(Dentist) * C(Method)', data=subset_data).fit()
    anova_table = sm.stats.anova_lm(model, typ=2)
    
    # Display ANOVA results
    print(anova_table)
```

### Insights:

- P value less than 0.05 for Dentist : method indicates a significant interaction seen for Alloytype 1.
- P value more than 0.05 for Dentist : method indicates no significant interaction seen for Alloytype 2.


### 4.4 How does the hardness of implants vary depending on dentists and methods together?


- Since we are asked to compare the mean hardness of implants between two factors, i.e., dentist and method together, it is a two way ANOVA test.
- The pairwise_tukeyhsd() function of Statsmodels will be used to compute the test statistic and p-value.

***Let's write the null and alternative hypothesis***

>$H_0$: No significant interaction effect between dentists and methods with respect to alloy type\
>$H_a$: Significant interaction effect observed between dentists and methods with respect to alloy type

### Checking the assumptions:

- Assumption1 - For testing of normality, Shapiro-Wilk’s test is applied to the response variable - this is already performed in 4.1 question
- Assumption2 - For equality of variance, Levene test is applied to the response variable 'Methods' and 'Dentist' which is already performed in the 4.2 question
- Assumption3 - Normally distributed data

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Creating an interaction group of both dentist and method
dental['Interaction'] = dental['Dentist'] + dental['Method']

# Perform Tukey HSD test on interaction groups
m_comp = pairwise_tukeyhsd(endog=dental['Response'], groups=dental['Interaction'], alpha=0.05)

# Print the results
print(m_comp)
```

### Insight: 
- 2-8, 3-8,4-8,5-8 do not show effects as the p value is high, so dentist and method may not impact hardness for these combinations.
- 2-3,2-7,4-7 show low p-values, which indicates dentist and method influence the hardness.
- This suggests that dentist and some methods together influence hardness largely when compared to some other methods.
